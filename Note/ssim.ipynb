```python
import cv2
import numpy as np
from skimage.metrics import structural_similarity as ssim
import matplotlib.pyplot as plt

#img1=np.zeros((300,300)).astype(np.uint8)
#img1[0:300,0:300]=255
#print(img1.dtype)
#plt.figure()
#plt.imshow(img1,cmap='gray')

img1 = cv2.imread(r'C:\Users\MelvinTheArch\Desktop\Research\AIE 2D\\test_repro\\DoC_98.png', cv2.IMREAD_GRAYSCALE)
img2 = cv2.imread(r'C:\Users\MelvinTheArch\Desktop\Research\AIE 2D\\circle.png', cv2.IMREAD_GRAYSCALE)
img3=img1-img2
print(img3.max(),img3.min())
#print(img1.dtype)
# 1. Create masks for non-black pixels
threshold = 1  # tolerance for "near-black" pixels
mask1 = img1 > threshold
mask2 = img2 > threshold

# 2. Union mask: any pixel that's foreground in EITHER image
union_mask = mask1 | mask2

# 3. Compute SSIM only on foreground region
score, ssim_map = ssim(img1, img2, full=True)

# 4. Extract SSIM values within the foreground mask only
foreground_ssim = ssim_map[union_mask].mean()
print(f"Full SSIM:       {score:.4f}")
print(f"Foreground SSIM: {foreground_ssim:.4f}")
```

```python
import numpy as np 
from PIL import Image 


def zoom_at(img, x, y, zoom):
    w, h = img.size
    zoom2 = zoom * 2
    img = img.crop((x - w / zoom2, y - h / zoom2, 
                    x + w / zoom2, y + h / zoom2))
    return img

#img_og=Image.open('./260812/260812_30mW_0mMol_Sync_rect_InterOpt/aaa_final_opt_mask.png')
#img_og=np.array(img_og)

#og_zoomed=zoom_at(img_og, 380,240, 10)

img_opt=Image.open('./260812/260812_30mW_0mMol_Sync_rect_TestOpt/aaa_final_opt_mask.png')
img_opt_np=np.array(img_opt)
w,h=img_opt.size
#img_opt=np.array(img_opt)

#og_zoomed.save('./260805/260805_30mW_0mMol_Stairs_Simulation/aaa_final_opt_mask_zoomed.png')
opt_zoomed=zoom_at(img_opt, 390,220, 40)
opt_cropped=img_opt.crop((w/2+70-5,h/2-50-5,w/2+70+5,h/2+5-50))
opt_cropped_np=np.array(opt_cropped)
print(opt_cropped_np)
np.savetxt('./260812/260812_30mW_0mMol_Sync_rect_TestOpt/aaa_final_opt_mask_cropped.txt',opt_cropped_np,fmt='%d')
np.savetxt('./260812/260812_30mW_0mMol_Sync_rect_TestOpt/fullmask.txt',img_opt_np,fmt='%d')
#opt_zoomed=np.array(opt_zoomed,dtype=np.uint8)
#print(opt_zoomed[16,16:])
opt_cropped
opt_cropped.save('./260812/260812_30mW_0mMol_Sync_rect_TestOpt/aaa_final_opt_mask_cropped.png')
```

```python
import numpy as np 
from PIL import Image 


def zoom_at(img, x, y, zoom):
    w, h = img.size
    zoom2 = zoom * 2
    img = img.crop((x - w / zoom2, y - h / zoom2, 
                    x + w / zoom2, y + h / zoom2))
    return img

img_og=Image.open('./260805/260805_30mW_0mMol_Stairs_Simulation/aaa_final_opt_mask.png')
#img_og=np.array(img_og)

og_zoomed=zoom_at(img_og, 380,240, 10)

img_opt=Image.open('./260805/260805_30mW_0mMol_Stairs_Simulation_opt/aaa_final_opt_mask.png')
#img_opt=np.array(img_opt)

og_zoomed.save('./260805/260805_30mW_0mMol_Stairs_Simulation/aaa_final_opt_mask_zoomed.png')
opt_zoomed=zoom_at(img_opt, 410,240, 10)
#opt_zoomed=np.array(opt_zoomed,dtype=np.uint8)
#print(opt_zoomed[16,16:])
opt_zoomed
opt_zoomed.save('./260805/260805_30mW_0mMol_Stairs_Simulation_opt/aaa_final_opt_mask_zoomed.png')




```
